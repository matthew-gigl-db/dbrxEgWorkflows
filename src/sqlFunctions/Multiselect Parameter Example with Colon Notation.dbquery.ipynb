```sql
-- results with hard coded filter values
FROM mgiglia.hls_genie.providers |> 
WHERE city in ('WORCESTER', 'BOSTON', 'SPRINGFIELD') |>
AGGREGATE count(provider_id) as provider_cnt GROUP BY city |>
ORDER BY provider_cnt DESC;

-- how the multiselect parameter is passed using the colon notation
select :cities as raw_parameter_value;

-- query with in statement using multiselect parameter using the colon notation
FROM mgiglia.hls_genie.providers |> 
WHERE city in (:cities) |>
AGGREGATE count(provider_id) as provider_cnt GROUP BY city |>
ORDER BY provider_cnt DESC;

-- declare a variable for cities as an array and pass split the multiselect parameter
declare or replace variable cities array<string>;
set variable cities = split(:cities, ",");
select cities;

-- then use array_contains to filter the cities
FROM mgiglia.hls_genie.providers |> 
WHERE array_contains(cities, city) |>
AGGREGATE count(provider_id) as provider_cnt GROUP BY city |>
ORDER BY provider_cnt DESC;
```
